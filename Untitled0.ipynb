```python
!pip install -q earthengine-api pandas
"""
Étape 2a - Extraction de la pluie CHIRPS par région du Niger (Google Earth Engine).

À exécuter dans Google Colab (ou en local avec earthengine-api installé).
Sortie : data/chirps_regions.csv  (une ligne par région et par année)

Règle anti-fuite d'information : la prédiction se fait à la mi-août.
Les variables de septembre et octobre sont extraites pour l'exploration
seulement ; elles NE DOIVENT PAS servir de variables d'entrée du modèle.

Version 2 : une requête par année (au lieu d'une seule requête géante), avec
toutes les variables regroupées dans une image multi-bandes, pour éviter l'erreur
« Too many concurrent aggregations ».
"""
import os
import time
import unicodedata

import ee
import pandas as pd

# ------------------------------------------------------------------ paramètres
PROJECT_ID = "credible-bridge-513611-c2"   # <- identifiant de ton projet Google Cloud enregistré
YEARS = range(2015, 2025)                  # CHIRPS dispo bien avant ; les rendements, eux, commencent en 2020
MONTHS = range(5, 11)                      # mai à octobre
SCALE = 5566                               # résolution native de CHIRPS (0,05 degré), en mètres
OUT_CSV = "data/chirps_regions.csv"

# ------------------------------------------------------------------ connexion
try:
    ee.Initialize(project=PROJECT_ID)
except Exception:
    ee.Authenticate()                      # ouvre une fenêtre de connexion Google (une seule fois)
    ee.Initialize(project=PROJECT_ID)

# ------------------------------------------------------------------ régions
# Limites administratives de niveau 1 (régions) du Niger.
# Remarque : cette couche est signalée « dépréciée » (remplacée par FAO/GAUL/2025/level1)
# mais elle fonctionne ; le rapport devra citer la version utilisée (GAUL 2015).
regions_fc = (
    ee.FeatureCollection("FAO/GAUL/2015/level1")
    .filter(ee.Filter.eq("ADM0_NAME", "Niger"))
    .select(["ADM1_NAME"])
)
noms_gee = regions_fc.aggregate_array("ADM1_NAME").getInfo()
print("Régions trouvées dans la couche de limites :", noms_gee)
assert len(noms_gee) == 8, "On attend 8 régions : vérifier la couche de limites (voir message ci-dessus)."

chirps = ee.ImageCollection("UCSB-CHG/CHIRPS/DAILY").select("precipitation")


def image_annee(y):
    """Image multi-bandes pour l'année y : une bande par variable."""
    bandes = []
    # Cumuls mensuels (mm)
    for m in MONTHS:
        debut = ee.Date.fromYMD(y, m, 1)
        cumul = chirps.filterDate(debut, debut.advance(1, "month")).sum()
        bandes.append(cumul.rename(f"pluie_mois_{m:02d}_mm"))
    # Variables "mi-saison" : du 1er mai au 15 août inclus (la date de fin de filterDate est exclue)
    saison = chirps.filterDate(ee.Date.fromYMD(y, 5, 1), ee.Date.fromYMD(y, 8, 16))
    bandes.append(saison.sum().rename("pluie_cumul_1mai_15aout_mm"))
    # Nombre de jours de pluie (> 1 mm) sur la même période
    bandes.append(saison.map(lambda im: im.gt(1)).sum().rename("jours_pluie_1mai_15aout"))
    return ee.Image.cat(bandes)


def extraire(y, essais=5):
    """Une requête par année, avec nouvelles tentatives en cas de limite de débit (erreur 429)."""
    fc = image_annee(y).reduceRegions(collection=regions_fc, reducer=ee.Reducer.mean(), scale=SCALE)
    for k in range(essais):
        try:
            return fc.getInfo()["features"]
        except ee.EEException as e:
            if k == essais - 1:
                raise
            attente = 10 * (k + 1)
            print(f"  limite de débit ({e}); nouvel essai dans {attente} s")
            time.sleep(attente)


# ------------------------------------------------------------------ extraction
lignes = []
for y in YEARS:
    print(f"Année {y}...")
    for f in extraire(y):
        p = dict(f["properties"])
        p["annee"] = y
        lignes.append(p)
    time.sleep(2)                          # petite pause entre deux années

df = pd.DataFrame(lignes).rename(columns={"ADM1_NAME": "region_gee"})


# ------------------------------------------------------------------ noms de régions harmonisés
def normaliser(s):
    return unicodedata.normalize("NFKD", s).encode("ascii", "ignore").decode().lower().strip()


CORRESPONDANCE = {
    "agadez": "Agadez", "diffa": "Diffa", "dosso": "Dosso", "maradi": "Maradi",
    "niamey": "Niamey", "tahoua": "Tahoua", "tillaberi": "Tillabéri", "zinder": "Zinder",
}
df["region"] = df["region_gee"].map(lambda s: CORRESPONDANCE.get(normaliser(s)))
assert df["region"].notna().all(), f"Noms non reconnus : {df.loc[df.region.isna(), 'region_gee'].unique()}"

# ------------------------------------------------------------------ mise en forme et sauvegarde
colonnes_vars = [c for c in df.columns if c.startswith(("pluie_", "jours_"))]
large = df[["region", "annee"] + sorted(colonnes_vars)].round(1).sort_values(["region", "annee"])

os.makedirs(os.path.dirname(OUT_CSV), exist_ok=True)
large.to_csv(OUT_CSV, index=False)
print(f"\n{len(large)} lignes écrites dans {OUT_CSV}")

# ------------------------------------------------------------------ contrôles rapides
print("\nPluie cumulée 1er mai - 15 août (mm), moyenne 2015-2024 par région :")
print(large.groupby("region")["pluie_cumul_1mai_15aout_mm"].mean().round(0).sort_values())
print("\nValeurs manquantes par colonne :")
print(large.isna().sum())
```
